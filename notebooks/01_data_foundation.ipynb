PHASE1: DATA FOUNDATION

```python
# ============================================================
# STEP 1 — Import and Construct Dataset
# ============================================================

import pandas as pd
import numpy as np

# importing all dataframes of each currencies, saving in seperate dfs
df_CNY = pd.read_csv(r"D:\Github Repositories\FX-Risk-Early-Warning\data\raw\China, CNYUSD.csv")
df_EUR = pd.read_csv(r"D:\Github Repositories\FX-Risk-Early-Warning\data\raw\Euro area, EURUSD.csv")
df_INR = pd.read_csv(r"D:\Github Repositories\FX-Risk-Early-Warning\data\raw\India, INRUSD.csv")
df_ZAR = pd.read_csv(r"D:\Github Repositories\FX-Risk-Early-Warning\data\raw\South Africa, ZARUSD.csv")
df_TRY = pd.read_csv(r"D:\Github Repositories\FX-Risk-Early-Warning\data\raw\Türkiye, TRYUSD.csv")
df_JPY = pd.read_csv(r"D:\Github Repositories\FX-Risk-Early-Warning\data\raw\Japan, JPYUSD.csv")
df_GBP = pd.read_csv(r"D:\Github Repositories\FX-Risk-Early-Warning\data\raw\United Kingdom, GBPUSD.csv")
df_BRL = pd.read_csv(r"D:\Github Repositories\FX-Risk-Early-Warning\data\raw\Brazil, BRLUSD.csv")
df_AUD = pd.read_csv(r"D:\Github Repositories\FX-Risk-Early-Warning\data\raw\Australia, AUDUSD.csv")

# forming main dataframe: df_fx_curr
dfs = [df_CNY, df_EUR, df_INR, df_ZAR, df_TRY, df_JPY, df_GBP, df_BRL, df_AUD]

df_fx_curr = dfs[0]

for df in dfs[1:]:
    df_fx_curr = pd.merge(df_fx_curr, df, on="Day", how = "inner")

# freezeing raw data
df_raw = df_fx_curr.copy()
df_fx_curr = df_raw.copy()

# ============================================================
# STEP 2 — Data Quality
# ============================================================

# setting the types of values in each column in df
df_fx_curr["Day"] = pd.to_datetime(df_fx_curr["Day"])
df_fx_curr.iloc[:, 1:] = df_fx_curr.iloc[:, 1:].astype(float)

# Sort chronologically
df_fx_curr = df_fx_curr.sort_values("Day").reset_index(drop=True)

# Basic data-quality diagnostics
print("Missing values:")
print(df_fx_curr.isna().sum())

print("\nDuplicated dates:")
print(df_fx_curr[df_fx_curr["Day"].duplicated()])

print("\nNegative values:")
print((df_fx_curr.iloc[:, 1:] < 0).sum())

print("\nShape:")
print(df_fx_curr.shape)

print("\nDescriptive statistics:")
print(df_fx_curr.describe())

print("\nData information:")
print(df_fx_curr.info())

print("\nFirst 5 rows:")
print(df_fx_curr.head())

# ============================================================
# SAVE PROCESSED DATASET
# ============================================================

import os

os.makedirs("../data/processed", exist_ok=True)

df_fx_curr.to_csv("../data/processed/fx_dataset_clean.csv", index=False)
```
